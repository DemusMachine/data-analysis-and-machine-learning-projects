```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

df = pd.read_csv('mail_data.csv')
df.head(3)
```

```python
data = {'ham':0,'spam':1}
df['Category_Encoded'] = df['Category'].map(data)
```

```python
df.drop_duplicates(inplace=True)
df_new = df.drop('Category', axis = 1)
df_new
```

```python
df_new.reset_index(drop=True, inplace=True)
df_new
```

```python
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df_new['Message'])
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, df_new['Category_Encoded'], test_size = 0.2)
```

```python
model = MultinomialNB()
model.fit(X_train,y_train)
```

```python
yPred = model.predict(X_test)
accuracy = accuracy_score(y_test,yPred)
print(accuracy)
```

```python
def predict(message):
    messageVector = vectorizer.transform([message])
    prediction = model.predict(messageVector)
    return 'Spam' if prediction[0] == 1 else 'Ham'

inp = input("Please enter email to predict: ")
print(f"The email is likely to be: {predict(inp)}")
```

```python
from sklearn.metrics import precision_score
precision = precision_score(y_test,yPred)
print(precision)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score, classification_report

print(classification_report(y_test, yPred))
```
